```python
import os
from ultralytics import YOLO

model = YOLO('yolov8m-seg.pt')

dataset_path = dataset.location
data_yaml_path = os.path.join(dataset_path, 'data.yaml')
print(f"Dataset located at: {dataset_path}")
print(f"Data YAML path: {data_yaml_path}")
```

```python
results = model.train(
    data=data_yaml_path,
    epochs=40,
    patience=5,
    imgsz=640,
    batch=16,
    workers=2
)
```

```python
metrics = model.val()

print(f"Mean Average Precision (mAP) 50-95 (Box): {metrics.box.map:.4f}")
print(f"Mean Average Precision (mAP) 50-95 (Mask): {metrics.seg.map:.4f}")
```
